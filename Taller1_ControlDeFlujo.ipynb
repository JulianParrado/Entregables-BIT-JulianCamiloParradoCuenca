### 1. Registro de estudiantes
Crea una lista llamada estudiantes.
Cada elemento de la lista debe ser una tupla con la siguiente estructura:

(nombre, edad, nota)
La lista debe contener mínimo 8 estudiantes.

```python
estudiantes = [
    ("Juan", 20, 4.6),
    ("María", 22, 4.4),
    ("Pedro", 19, 3.2),
    ("Ana", 21, 4.9),
    ("Luis", 23, 3.5),
    ("Carla", 20, 2.9),
    ("Julian", 22, 4.8),
    ("Fernanda", 21, 3.7)
]
```

### 2. Mostrar la información
Utiliza un ciclo for para recorrer la lista e imprimir la información de cada estudiante con un formato similar al siguiente:

Ana tiene 20 años y obtuvo una nota de 4.5

```python
for nombre, edad, nota in estudiantes:
    print(f"{nombre} tiene {edad} y obtuvo una nota de {nota}")
```

### 3. Clasificación de estudiantes
Utilizando estructuras if, elif y else, clasifica a cada estudiante según su nota.

| Nota | Clasificación |
|------|---------------|
| Mayor o igual a 4.5 | Excelente |
| Entre 4.0 y 4.49 | Bueno |
| Entre 3.0 y 3.99 | Aceptable |
| Menor a 3.0 | Reprobó |

Imprime el nombre del estudiante junto con su clasificación.

```python
for nombre, edad, nota in estudiantes:
    if nota >= 4.5:
        print(f"{nombre}: Excelente")
    elif nota >= 4.0:
        print(f"{nombre}: Bueno")
    elif nota >= 3.0:
        print(f"{nombre}: Aceptable")
    else:
        print(f"{nombre}: Reprobó")
```

### 4. Promedio general
Calcula el promedio de todas las notas utilizando un ciclo y una variable acumuladora.
Al finalizar imprime el promedio obtenido.

```python
suma = 0

for nombre, edad, nota in estudiantes:
    suma += nota

print(f"El promedio de notas fue de {suma/len(estudiantes):.1f}")
```

### 5. Búsqueda de estudiante
Solicita al usuario el nombre de un estudiante mediante input().

Si el estudiante existe en la lista, imprime un mensaje indicando que fue encontrado.

Si no existe, informa que no se encontró ningún estudiante con ese nombre.

```python
busqueda = input("Nombre del estudiante: ")
val = False

for nombre, edad, nota in estudiantes:
    if busqueda.lower() == nombre.lower():
        val = True
        print(f"Estudiante {nombre} encontrado")
        break

if not val:
    print(f"Estudiante {busqueda} no encontrado")
```

### 6. Diccionario de ciudades
Crea un diccionario donde:

    • La llave sea el nombre del estudiante. 

    • El valor sea la ciudad donde vive. 
    
Recorre el diccionario utilizando un ciclo e imprime la información con el siguiente formato:

Ana vive en Bogotá

```python
est_ciudad = {
    "Juan": "Bogotá",
    "Jorge": "Manizalez",
    "Camila": "Villavicencio",
    "Ana": "Pereira",
    "Luis": "Bogotá",
    "María": "Bogotá",
    "Enrique": "Pereira",
    "Viviana": "Manizalez",
    "Armando": "Medellín",
    "Nicolás": "Medellín",
    "Ivan": "Medellín"
}
```

Los diccionarios tienen tres métodos principales para recorrerse:

.keys() Recorre solo las llaves (nombres)

.values() Recorre solo los valores (ciudades)

.items() Recorre pares (llave, valor) al mismo tiempo.

```python
for nom, ciud in est_ciudad.items():
    print(f"{nom} vive en {ciud}")
```

### 7. Cantidad de estudiantes por ciudad
A partir del diccionario anterior, construye un nuevo diccionario que indique cuántos estudiantes viven en cada ciudad.
Ejemplo:

{

    "Bogotá": 3,

    "Medellín": 2,

    "Pereira": 1
    
}

```python
ciudad_count = {}

for ciud in est_ciudad.values():
    if ciud not in ciudad_count:
        ciudad_count[ciud] = 1
    else:
        ciudad_count[ciud] += 1

print(ciudad_count)

```

### 8. Ciclo While
Construye un programa que permita ingresar números utilizando un ciclo while.

El programa debe finalizar únicamente cuando el usuario ingrese el número 0.

Al finalizar muestra:

    • Cantidad de números ingresados (sin contar el 0). 
    • Suma total. 
    • Promedio de los números ingresados. 

```python
num = []

while True:
    numero = input("Ingrese número: ")
    if numero == '0':
        break
    num.append(float(numero))

print(f"Cantidad de números: {len(num)}")
print(f"Suma total: {sum(num)}")
print(f"Promedio: {sum(num)/len(num):.2f}" if num else "No se ingresaron números")
print(f"Los números fueron: {num}")
    
```

### 9. Uso de break y continue
Recorre los números del 1 al 30.

    • Si el número es múltiplo de 3, utiliza continue. 
    • Si el número es igual a 25, utiliza break. 
    
Imprime únicamente los números que fueron procesados por el ciclo.

```python
process = []

for i in range(1,31):
    if i == 25:
        break
    elif i%3 == 0:
        continue

    process.append(i)

print(process)
        
```
