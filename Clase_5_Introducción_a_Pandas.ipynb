## Qué es Pandas?

- Biblioteca de Python para manipulación y análisis de datos

- Trabaja con estructuras de datos tipo tablas (DataFrames)

- Ideal para trabajar con datos estructurados (CSV, Excel, SQL, etc)

### Conceptos Clave:

- Series: Arreglo unidimensional etiquetado

- DataFrame: Estructura bidimensional tipo tabla

- Índices: Etiquetas para acceder a los datos

## Series vs Dataframe

```python
# Una serie es una columna

edades = pd.Series([20,25,30])

edades
```

```python
# Un Dataframe es una tabla

datos = {
    "Nombre":["Ana","Luis","Pedro"],
    "Edad":[20,25,30],
    "Ciudad":["Bogotá","Pereira","Cali"]
}

df = pd.DataFrame(datos)

df
```
